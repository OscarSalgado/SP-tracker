# Kolmogorov-Arnold Networks for Bearing Fault Diagnosis

**ArXiv**: [2412.01322](https://arxiv.org/abs/2412.01322)

```python
import numpy as np
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import sys
sys.path.insert(0, '.')

from src import BearingFaultDiagnoser

print("✓ Librerías cargadas")
```

```python
# Generar señal de vibración sintética
t = np.arange(0, 2, 0.0001)
bearing_freq = 60  # Hz
impulse_freq = 120  # Hz

signal = (
    0.3 * np.sin(2 * np.pi * bearing_freq * t) +
    0.2 * np.sin(2 * np.pi * impulse_freq * t) +
    0.15 * np.random.randn(len(t))
)

print(f"Señal: {len(signal)} muestras, {signal.std():.4f} std")
```

```python
diagnoser = BearingFaultDiagnoser(input_size=64)
result = diagnoser.diagnose(signal)

print(f"Diagnóstico: {result['fault_name']}")
print(f"Clase: {result['fault_class']}")
print(f"Características: {len(result['features'])} features")
print("OK")
```
