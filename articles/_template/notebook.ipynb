# Plantilla de artículo — reproducción y visualización

Ejemplo mínimo: carga el código de `src/`, genera datos sintéticos triviales y visualiza el resultado.

```python
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

from src import square

x = list(range(-5, 6))
y = [square(v) for v in x]
y
```

```python
fig, ax = plt.subplots()
ax.plot(x, y, marker="o")
ax.set_xlabel("x")
ax.set_ylabel("square(x)")
ax.set_title("Resultado reproducido")
```
